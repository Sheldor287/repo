```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import numpy as np
df = pd.read_stata('arrests-latest.dta')
state_abbrev = {
    'ALABAMA':'AL','ALASKA':'AK','ARIZONA':'AZ','ARKANSAS':'AR','CALIFORNIA':'CA',
    'COLORADO':'CO','CONNECTICUT':'CT','DELAWARE':'DE','FLORIDA':'FL','GEORGIA':'GA',
    'HAWAII':'HI','IDAHO':'ID','ILLINOIS':'IL','INDIANA':'IN','IOWA':'IA','KANSAS':'KS',
    'KENTUCKY':'KY','LOUISIANA':'LA','MAINE':'ME','MARYLAND':'MD','MASSACHUSETTS':'MA',
    'MICHIGAN':'MI','MINNESOTA':'MN','MISSISSIPPI':'MS','MISSOURI':'MO','MONTANA':'MT',
    'NEBRASKA':'NE','NEVADA':'NV','NEW HAMPSHIRE':'NH','NEW JERSEY':'NJ','NEW MEXICO':'NM',
    'NEW YORK':'NY','NORTH CAROLINA':'NC','NORTH DAKOTA':'ND','OHIO':'OH','OKLAHOMA':'OK',
    'OREGON':'OR','PENNSYLVANIA':'PA','RHODE ISLAND':'RI','SOUTH CAROLINA':'SC','SOUTH DAKOTA':'SD',
    'TENNESSEE':'TN','TEXAS':'TX','UTAH':'UT','VERMONT':'VT','VIRGINIA':'VA','WASHINGTON':'WA',
    'WEST VIRGINIA':'WV','WISCONSIN':'WI','WYOMING':'WY','DISTRICT OF COLUMBIA':'DC'
}

# Count occurrences
states = df['apprehension_state']
state_counts = states.map(state_abbrev).value_counts().reset_index()
state_counts.columns = ['state', 'count']

# Take log (log1p to handle 0)
state_counts['log_count'] = np.log1p(state_counts['count'])

# Plot
fig = px.choropleth(
    state_counts,
    locations='state',
    locationmode='USA-states',
    color='log_count',
    scope='usa',
    color_continuous_scale='Reds',
    labels={'log_count': 'log(count + 1)'}
)
fig.show()
```

```python


# Count occurrences
state_2025 = df[df['apprehension_date'] >= '2025-03-01']['apprehension_state']
state_counts2025 = state_2025.map(state_abbrev).value_counts().reset_index()
state_counts2025.columns = ['state', 'count']

# Take log (log1p to handle 0)
state_counts2025['log_count'] = np.log1p(state_counts2025['count'])

# Plot
fig = px.choropleth(
    state_counts2025,
    locations='state',
    locationmode='USA-states',
    color='log_count',
    scope='usa',
    color_continuous_scale='Reds',
    labels={'log_count': 'log(count + 1)'}
)
fig.show()
```

```python
compare = state_counts.compare(state_counts2025)
df_C = pd.merge(state_counts, state_counts2025, on='state', how='outer', suffixes=('_total', '_2025')).fillna(0)
df_C['diff'] = -np.log1p(df_C['count_2025']) + np.log1p(df_C['count_total'])
new_count = df_C['diff']
# Plot
fig = px.choropleth(
    df_C,
    locations='state',
    locationmode='USA-states',
    color='diff',
    scope='usa',
    color_continuous_scale='Reds',
    labels={'log_count': 'log(count + 1)'}
)
fig.show()
```
