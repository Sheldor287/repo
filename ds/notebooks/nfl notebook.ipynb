```python
import pandas as pd
from nflreadpy import load_schedules, load_team_stats

# Load
games = load_schedules(seasons=range(2010, 2024)).to_pandas()
team_stats = load_team_stats(seasons=range(2010, 2024)).to_pandas()



# Merge on season + week
games_small = games[["season", "week", "home_team", "away_team", "home_score", "away_score"]]
team_long = team_stats.merge(
    games_small,
    on=["season", "week"],
    how="left"
)
team_long["home"] = (team_long["team"] == team_long["home_team"]).astype(int)
# Reorder columns
home_col = team_long.pop("home")
team_long.insert(team_long.columns.get_loc("team") + 1, "home", home_col)
team_long = team_long.drop(columns=["home_team", "away_team"])




team_long



```
