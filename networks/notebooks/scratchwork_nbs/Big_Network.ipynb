```python
import networkx as nx
import matplotlib.pyplot as plt

def read_mtx(file='geo1k_14k.mtx'):
    G = nx.DiGraph()
    with open(file) as f:
        for line in f:
            if line.startswith('%'):
                continue
            parts = line.split()
            if len(parts) == 2:
                i, j = map(int, parts)
                G.add_edge(i-1, j-1)  # convert to 0-index
    return G


G = read_mtx()
print(f"Number of nodes: {G.number_of_nodes()}")
```

```python
pos = nx.kamada_kawai_layout(G)
plt.figure(figsize=(10,10))
nx.draw(G,pos,node_size=2,width=0.4,with_labels=False)
plt.show()
# time: 65
```
