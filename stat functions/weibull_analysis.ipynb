```python
import numpy as np
import pandas as pd
import scipy.stats as stats

def create_data(events: int) -> pd.DataFrame:
    np.random.seed(123)
    true_k = 1.8
    true_lmda = 150
    mock_gaps = true_lmda * np.random.weibull(true_k, events)

    start_date = pd.to_datetime("2025-01-01")
    failure_dates = []
    current_date = start_date

    for gap in mock_gaps:
        current_date += pd.to_timedelta(gap, unit='D')
        failure_dates.append(current_date)
    
    df_calendar = pd.DataFrame({
        'event': [f"Failure_{i}" for i in range(len(failure_dates))],
        'calendar_date': failure_dates
    })
    return df_calendar

def transform_dates_to_gaps(df: pd.DataFrame, date_col: str, gap_col: str, start_date: pd.Timestamp) -> pd.DataFrame:
    df[gap_col] = df[date_col].diff().dt.total_seconds()/(24*3600)
    df.loc[0, gap_col] = (df.loc[0, date_col] - start_date).total_seconds()/(24*3600)
    return df

def weibull_math(df, gap_col):
    gap_array = df[gap_col].values
    params = stats.weibull_min.fit(gap_array, floc=0)
    k, loc, lmda = params
    

```
