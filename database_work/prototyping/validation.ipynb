```python
from faker import Faker
import pandas as pd
import random 

def phone_number():
    num_list = [random.randint(0, 9) for i in range(10)]
    num_list.insert(6, "-")
    num_list.insert(3, "-")
    return "".join(map(str,num_list))

fake = Faker()
date = fake.date_time_this_year()

df = pd.DataFrame({
    "uuid": [fake.uuid4() for _ in range(10)],
    "name": [fake.name() for _ in range(10)],
    "phone": [phone_number() for _ in range(10)],
    "job": [fake.job() for _ in range(10)],
    "vin": [fake.vin() for _ in range(10)],
    "city": [fake.city() for _ in range(10)],
    "governance_status": ["DRAFT" for _ in range(10)],
    "created_at": [date for _ in range(10)]
})
```

```python
print(f"columns: {list(df.columns)}")
print(f"Row 0:\n{df.iloc[0]}")
```
