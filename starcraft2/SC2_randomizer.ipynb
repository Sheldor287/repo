```python
import numpy as np

def double_pick(picks):
    return [f'({picks[i]}, {picks[i+1]})' for i in range(0, len(picks), 2)]

people = ['Sheldor', 'MasterMind', 'Global', 'Zing']
units = [
    'marine', 'marauder', 'reaper', 'ghost',
    'helion', 'mine', 'banshee', 'BC', 
    'tank', 'viking', 'thor', 'cyclone', 'lib', 'WILD',

    'ling', 'roach', 'ravager','hydra', 'bane', 
    'queen', 'muta', 'corruptor', 'lurker', 
    'ultra', 'swarmhost', 'WILD',

    'zealot', 'adept', 'centry', 'stalker', 'archon',
    'dt', 'WILD', 'immortal', 'colossus', 
    'disruptor', 'phoenix', 'voidray', 'oracle', 'tempest', 
]
weights = [
    4, 2, 2, 2, 
    3, 2, 2, 1, 
    4, 2, 1, 4, 1, 2, 

    5, 4, 2, 3, 2, 
    2, 4, 4, 1, 
    0.5, 1.825, 0.125, 
    
    3, 3, 3, 5, 2,
    2, 1, 2, 0.5, 
    0.5, 4, 3, 2, 2, 
    ]

terran_picks = np.random.choice(units[0:14], size=len(2*people), replace=False, p=np.array(weights[0:14])/sum(weights[0:14]))#, weights=weights[0:14])
zerg_picks = np.random.choice(units[14:25], size=len(2*people), replace=False, p=np.array(weights[14:25])/sum(weights[14:25]))#, weights=weights[14:25])
protoss_picks = np.random.choice(units[25:], size=len(2*people), replace=False, p=np.array(weights[25:])/sum(weights[25:]))#, weights=weights[25:])
double=False

if double: 
    terran_picks = double_pick(terran_picks)
    zerg_picks = double_pick(zerg_picks)
    protoss_picks = double_pick(protoss_picks)

for p, t, z, pr in zip(people, terran_picks, zerg_picks, protoss_picks):
    if p in ['Zing', 'Sammo',]:
        print(f'{p}: \n{t}    \n')
    elif p in ['Hail']:
        print(f'{p}: \n{pr}    \n')
    elif p in ['Eclypse']:
        print(f'{p}: \n{z}    \n')
    
    else:
        print(f'{p}: \n{t}, {z}, {pr}    \n')
```
