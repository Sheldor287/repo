````python

from zipfile import ZipFile
from pathlib import Path
from collections import Counter, defaultdict
from urllib.parse import urlparse, parse_qs
import xml.etree.ElementTree as ET
import hashlib



# Configuration
mdzip_path = Path("SAF_SCM_Profile.mdzip")
resource_name = "com.nomagic.magicdraw.uml_model.shared_model"
output_dir = Path("mdzip_analysis")
output_dir.mkdir(exist_ok=True)
output_path = output_dir / "shared_model.xmi"


# ---------------------------------------------------------
# XML helpers
# ---------------------------------------------------------

XMI_NS = "http://www.omg.org/XMI"


def local_name(tag):
    """Remove an XML namespace from a tag."""
    return tag.split("}", 1)[-1] if "}" in tag else tag


def get_xmi_attribute(element, name):
    return element.attrib.get(f"{{{XMI_NS}}}{name}")


def get_element_type(element):
    """Return xmi:type if present, otherwise the XML tag."""
    return get_xmi_attribute(element, "type") or local_name(element.tag)


# ---------------------------------------------------------
# Extract resource
# ---------------------------------------------------------

with ZipFile(mdzip_path) as archive:
    names = set(archive.namelist())
    if resource_name not in names:
        raise FileNotFoundError(f"Resource not found in archive: {resource_name}")
    raw_xml = archive.read(resource_name)
    output_path.write_bytes(raw_xml)
print(f"Extracted: {resource_name}")
print(f"Size:      {len(raw_xml):,} bytes")
print(f"SHA-256:   {hashlib.sha256(raw_xml).hexdigest()}")
print(f"Saved to:  {output_path}")


# ---------------------------------------------------------
# Parse XML
# ---------------------------------------------------------

root = ET.fromstring(raw_xml)

elements = list(root.iter())

print("\nXML SUMMARY")
print("-" * 60)
print(f"Root tag:       {local_name(root.tag)}")
print(f"Total XML nodes: {len(elements):,}")


# ---------------------------------------------------------
# Count tags and metaclasses
# ---------------------------------------------------------

tag_counts = Counter(local_name(e.tag) for e in elements)
type_counts = Counter(get_element_type(e) for e in elements)

print("\nMOST COMMON XML TAGS")
print("-" * 60)

for tag, count in tag_counts.most_common(30):
    print(f"{tag:<40} {count:>8,}")


print("\nMOST COMMON ELEMENT TYPES")
print("-" * 60)

for element_type, count in type_counts.most_common(40):
    print(f"{element_type:<40} {count:>8,}")


# ---------------------------------------------------------
# Identify elements with XMI IDs
# ---------------------------------------------------------

id_records = []

for element in elements:

    xmi_id = get_xmi_attribute(element, "id")

    if xmi_id:
        id_records.append({
            "id": xmi_id,
            "tag": local_name(element.tag),
            "type": get_element_type(element),
            "name": element.attrib.get("name"),
        })


id_counts = Counter(record["id"] for record in id_records)

duplicate_ids = {
    xmi_id: count
    for xmi_id, count in id_counts.items()
    if count > 1
}

print("\nIDENTIFIER SUMMARY")
print("-" * 60)
print(f"Elements with XMI IDs: {len(id_records):,}")
print(f"Unique XMI IDs:        {len(id_counts):,}")
print(f"Duplicate IDs:         {len(duplicate_ids):,}")

if duplicate_ids:
    print("\nDuplicate IDs (first 20):")
    for xmi_id, count in list(duplicate_ids.items())[:20]:
        print(f"{xmi_id}: {count}")


# ---------------------------------------------------------
# Identify references
# ---------------------------------------------------------

href_records = []
idref_records = []

for element in elements:

    for attr_name, value in element.attrib.items():

        attr = local_name(attr_name)

        if attr == "href":
            href_records.append({
                "element": local_name(element.tag),
                "type": get_element_type(element),
                "href": value,
            })

        elif attr == "idref":
            idref_records.append({
                "element": local_name(element.tag),
                "idref": value,
            })


print("\nREFERENCE SUMMARY")
print("-" * 60)
print(f"href references:   {len(href_records):,}")
print(f"xmi:idref values:  {len(idref_records):,}")

print("\nSAMPLE HREF REFERENCES")

for record in href_records[:30]:
    print(
        f"{record['element']:<30} "
        f"{record['href']}"
    )


# ---------------------------------------------------------
# Identify likely relationship elements
# ---------------------------------------------------------

relationship_terms = {
    "Association",
    "Generalization",
    "Dependency",
    "Abstraction",
    "Realization",
    "InterfaceRealization",
    "Usage",
    "Connector",
    "Transition",
    "ControlFlow",
    "ObjectFlow",
    "InformationFlow",
    "Satisfy",
    "Verify",
    "Refine",
    "Trace",
    "DeriveReqt",
    "Allocate",
    "BindingConnector",
}


relationship_records = []

for element in elements:

    element_type = get_element_type(element)
    type_local = element_type.split(":")[-1]

    if type_local in relationship_terms:

        relationship_records.append({
            "id": get_xmi_attribute(element, "id"),
            "tag": local_name(element.tag),
            "type": element_type,
            "name": element.attrib.get("name"),
        })


print("\nLIKELY RELATIONSHIP TYPES")
print("-" * 60)

relationship_counts = Counter(
    record["type"]
    for record in relationship_records
)

for relationship_type, count in relationship_counts.most_common():
    print(f"{relationship_type:<40} {count:>8,}")

print(f"\nTotal identified relationships: {len(relationship_records):,}")


# ---------------------------------------------------------
# Top-level structure
# ---------------------------------------------------------

print("\nTOP-LEVEL STRUCTURE")
print("-" * 60)

for child in root:

    print(
        f"Tag={local_name(child.tag):<30} "
        f"Type={get_element_type(child):<30} "
        f"Name={child.attrib.get('name')}"
    )
````

2. Determine how references actually connect elements:

The following script checks:
1. Which xmi:idref values resolve to an ID in this resource.
2. Which href references point to another object in this resource, another resource, or an external metamodel.
3. Which attributes contain values that match an element ID.
4. The parent-child context of important relationship properties.

````python

from collections import Counter, defaultdict
from urllib.parse import urlparse, parse_qs
import xml.etree.ElementTree as ET

path = "mdzip_analysis/shared_model.xmi"
root = ET.parse(path).getroot()


def local_name(value):
    return value.split("}", 1)[-1] if "}" in value else value


def get_attr(element, name):
    for key, value in element.attrib.items():
        if local_name(key) == name:
            return value
    return None


# ---------------------------------------------------------
# 1. Build an ID index
# ---------------------------------------------------------

id_index = defaultdict(list)

for element in root.iter():
    xmi_id = get_attr(element, "id")

    if xmi_id:
        id_index[xmi_id].append(element)


print("\nID INDEX")
print("-" * 70)
print(f"Unique IDs: {len(id_index)}")
print(f"Duplicate IDs: {sum(len(v) > 1 for v in id_index.values())}")


# ---------------------------------------------------------
# 2. Resolve idref references
# ---------------------------------------------------------

idref_results = Counter()
unresolved_idrefs = []

for element in root.iter():

    idref = get_attr(element, "idref")

    if not idref:
        continue

    if idref in id_index:
        idref_results["local"] += 1
    else:
        idref_results["unresolved"] += 1
        unresolved_idrefs.append({
            "tag": local_name(element.tag),
            "idref": idref,
        })


print("\nIDREF RESOLUTION")
print("-" * 70)

for status, count in idref_results.items():
    print(f"{status:<20} {count}")

print("\nUnresolved IDREFs:")
for record in unresolved_idrefs[:30]:
    print(record)


# ---------------------------------------------------------
# 3. Classify href references
# ---------------------------------------------------------

href_results = Counter()
href_examples = defaultdict(list)

for element in root.iter():

    href = get_attr(element, "href")

    if not href:
        continue

    parsed = urlparse(href)
    fragment = parsed.fragment

    if href.startswith("#"):
        category = "same-resource fragment"

    elif href.startswith("local:"):
        category = "local project resource"

    elif href.startswith("PROJECT-"):
        category = "project resource"

    elif parsed.scheme in {"http", "https"}:
        category = "external URL/metamodel"

    else:
        category = "other"

    href_results[category] += 1

    if len(href_examples[category]) < 10:
        href_examples[category].append({
            "element": local_name(element.tag),
            "href": href,
            "fragment": fragment,
            "fragment_resolves_locally": fragment in id_index,
        })


print("\nHREF CLASSIFICATION")
print("-" * 70)

for category, count in href_results.most_common():
    print(f"{category:<35} {count}")

for category, examples in href_examples.items():

    print(f"\n{category} examples:")

    for example in examples:
        print(example)


# ---------------------------------------------------------
# 4. Find attributes that reference existing IDs
# ---------------------------------------------------------

reference_attributes = Counter()
reference_examples = defaultdict(list)

for element in root.iter():

    for key, value in element.attrib.items():

        attr = local_name(key)

        if attr in {"id", "ID", "idref", "href"}:
            continue

        if value in id_index:

            reference_attributes[attr] += 1

            if len(reference_examples[attr]) < 10:
                reference_examples[attr].append({
                    "element": local_name(element.tag),
                    "value": value,
                })


print("\nID-VALUED ATTRIBUTES")
print("-" * 70)

for attr, count in reference_attributes.most_common():

    print(f"\n{attr}: {count}")

    for example in reference_examples[attr]:
        print(f"  {example}")


# ---------------------------------------------------------
# 5. Inspect relationship property context
# ---------------------------------------------------------

relationship_properties = {
    "generalization",
    "memberEnd",
    "ownedEnd",
    "navigableOwnedEnd",
    "ownedAttribute",
    "type",
    "association",
    "annotatedElement",
    "constrainedElement",
    "client",
    "supplier",
    "source",
    "target",
}

print("\nRELATIONSHIP PROPERTY EXAMPLES")
print("-" * 70)

def inspect(element, ancestors):

    tag = local_name(element.tag)

    if tag in relationship_properties:

        print({
            "parent": ancestors[-1] if ancestors else None,
            "property": tag,
            "attributes": {
                local_name(k): v
                for k, v in element.attrib.items()
            },
            "children": [local_name(c.tag) for c in element],
        })


def walk(element, ancestors=None):

    if ancestors is None:
        ancestors = []

    inspect(element, ancestors)

    current = ancestors + [local_name(element.tag)]

    for child in element:
        walk(child, current)


walk(root)
````

````python

from __future__ import annotations

import csv
import json
import re
import zipfile

from collections import Counter, defaultdict
from pathlib import Path
from urllib.parse import parse_qs, urlparse
import xml.etree.ElementTree as ET


# ============================================================
# CONFIGURATION
# ============================================================

# Set this to the path of your original MDZIP archive.
ARCHIVE_PATH = Path("SAF_SCM_Profile.mdzip")

# Directory for generated reports.
OUTPUT_DIR = Path("mdzip_analysis/archive_inventory")

# Print a limited number of examples per reference category.
MAX_EXAMPLES = 15

# Attributes other than id/idref/href that should be investigated
# when their values match an ID in the archive.
REFERENCE_CANDIDATE_ATTRIBUTES = {
    "association",
    "type",
    "owner",
    "owningElement",
    "general",
    "source",
    "target",
    "client",
    "supplier",
    "annotatedElement",
    "constrainedElement",
}


# ============================================================
# GENERAL HELPERS
# ============================================================

def local_name(value: str) -> str:
    """Return the local name of an XML tag or attribute."""
    return value.split("}", 1)[-1] if "}" in value else value


def namespace_uri(value: str) -> str | None:
    """Return the namespace URI of an expanded XML name."""
    if value.startswith("{"):
        return value[1:].split("}", 1)[0]
    return None


def get_attribute(element, name):
    """Get an attribute by local name, ignoring its namespace."""
    for key, value in element.attrib.items():
        if local_name(key) == name:
            return value
    return None


def is_xml_candidate(data: bytes) -> bool:
    """Quick check before attempting XML parsing."""
    sample = data[:2048].lstrip(b"\xef\xbb\xbf \t\r\n")
    return sample.startswith(b"<")


def write_csv(path: Path, rows: list[dict], fieldnames=None):
    """Write dictionaries to CSV."""
    if not rows and not fieldnames:
        return

    if fieldnames is None:
        fieldnames = list(rows[0].keys())

    with path.open("w", newline="", encoding="utf-8-sig") as f:
        writer = csv.DictWriter(
            f,
            fieldnames=fieldnames,
            extrasaction="ignore",
        )
        writer.writeheader()
        writer.writerows(rows)


# ============================================================
# ARCHIVE INVENTORY
# ============================================================

def inventory_archive(zip_file):
    """
    Inventory all archive members and attempt to parse XML files.

    Resource identity is initially the exact ZIP member name.
    """
    inventory = []
    resources = {}

    for info in zip_file.infolist():
        if info.is_dir():
            continue

        member = info.filename
        data = zip_file.read(member)

        record = {
            "member": member,
            "size_bytes": info.file_size,
            "compressed_size": info.compress_size,
            "extension": Path(member).suffix.lower(),
            "is_xml_candidate": is_xml_candidate(data),
            "xml_parse_success": False,
            "root_tag": None,
            "xml_node_count": 0,
            "parse_error": None,
        }

        if is_xml_candidate(data):
            try:
                root = ET.fromstring(data)

                record["xml_parse_success"] = True
                record["root_tag"] = local_name(root.tag)
                record["xml_node_count"] = sum(
                    1 for _ in root.iter()
                )

                resources[member] = {
                    "member": member,
                    "data": data,
                    "root": root,
                    "node_count": record["xml_node_count"],
                }

            except ET.ParseError as exc:
                record["parse_error"] = str(exc)

        inventory.append(record)

    return inventory, resources


# ============================================================
# RESOURCE LOOKUP
# ============================================================

def build_resource_lookup(resources):
    """
    Build lookup tables for archive resource names.

    Exact resource names are preferred. Basename matching is
    retained as a fallback, but ambiguous matches are not resolved
    automatically.
    """
    exact = {}
    basenames = defaultdict(list)

    for member in resources:
        exact[member] = member
        basenames[Path(member).name].append(member)

    return exact, basenames


def resolve_resource_name(resource_name, exact, basenames):
    """
    Resolve a resource name to an archive member.

    Returns:
        (member, status)

    Status:
        exact_match
        basename_match
        ambiguous_basename
        not_found
    """
    if not resource_name:
        return None, "missing_resource_name"

    resource_name = resource_name.replace("\\", "/")

    if resource_name in exact:
        return exact[resource_name], "exact_match"

    if resource_name in basenames:
        matches = basenames[resource_name]

        if len(matches) == 1:
            return matches[0], "basename_match"

        return None, "ambiguous_basename"

    return None, "not_found"


# ============================================================
# ID INDEX
# ============================================================

def build_id_indexes(resources):
    """
    Build an ID index for each resource and a global index.

    xmi:id and unqualified id are treated as the same local
    identifier concept, while uppercase ID is retained separately.
    """
    resource_indexes = {}
    global_index = defaultdict(list)
    id_rows = []

    for member, resource in resources.items():
        root = resource["root"]
        local_index = defaultdict(list)

        for element in root.iter():
            source_id = get_attribute(element, "id")

            if source_id:
                local_index[source_id].append(element)

                global_index[source_id].append({
                    "resource": member,
                    "element": element,
                })

                id_rows.append({
                    "resource": member,
                    "source_id": source_id,
                    "tag": local_name(element.tag),
                    "namespace": namespace_uri(element.tag),
                    "name": get_attribute(element, "name"),
                    "identifier_kind": "id",
                })

            # Preserve uppercase ID as a separate identifier.
            uppercase_id = get_attribute(element, "ID")

            if uppercase_id:
                id_rows.append({
                    "resource": member,
                    "source_id": uppercase_id,
                    "tag": local_name(element.tag),
                    "namespace": namespace_uri(element.tag),
                    "name": get_attribute(element, "name"),
                    "identifier_kind": "ID",
                })

        resource_indexes[member] = local_index

    return resource_indexes, global_index, id_rows


# ============================================================
# XML ELEMENT CONTEXT
# ============================================================

def build_parent_maps(resources):
    """Build parent maps for each resource."""
    parent_maps = {}

    for member, resource in resources.items():
        parent_map = {}

        for parent in resource["root"].iter():
            for child in parent:
                parent_map[child] = parent

        parent_maps[member] = parent_map

    return parent_maps


def nearest_identified_ancestor(element, parent_map):
    """Find the closest ancestor (including self) with an ID."""
    current = element

    while current is not None:
        source_id = get_attribute(current, "id")

        if source_id:
            return source_id, local_name(current.tag)

        current = parent_map.get(current)

    return None, None


# ============================================================
# REFERENCE RESOLUTION
# ============================================================

def parse_href(href):
    """
    Parse an href into its resource component and fragment.

    Handles:
      #id
      resource_name#id
      PROJECT-id?resource=resource_name#id
      local:/PROJECT-id?resource=resource_name#id
      http://.../UML.xmi#Class
    """
    parsed = urlparse(href)

    fragment = parsed.fragment or None
    query = parse_qs(parsed.query)

    resource_parameter = query.get("resource", [None])[0]

    if resource_parameter:
        return {
            "fragment": fragment,
            "resource_name": resource_parameter,
            "href_kind": "project_resource",
            "project_identifier": parsed.path.lstrip("/"),
        }

    if href.startswith("#"):
        return {
            "fragment": fragment,
            "resource_name": None,
            "href_kind": "same_resource",
            "project_identifier": None,
        }

    if parsed.scheme in {"http", "https"}:
        return {
            "fragment": fragment,
            "resource_name": None,
            "href_kind": "external_url",
            "project_identifier": None,
        }

    if fragment:
        return {
            "fragment": fragment,
            "resource_name": parsed.path or None,
            "href_kind": "external_or_relative_resource",
            "project_identifier": None,
        }

    return {
        "fragment": None,
        "resource_name": parsed.path or None,
        "href_kind": "other",
        "project_identifier": None,
    }


def resolve_href(
    href,
    current_resource,
    resource_indexes,
    exact_lookup,
    basename_lookup,
    global_index,
):
    """Resolve an href against the archive's resource inventory."""
    parsed = parse_href(href)

    fragment = parsed["fragment"]
    kind = parsed["href_kind"]
    resource_name = parsed["resource_name"]

    result = {
        **parsed,
        "resolved_resource": None,
        "resolution_status": None,
        "target_count": 0,
    }

    if not fragment:
        result["resolution_status"] = "no_fragment"
        return result

    # A bare #fragment refers to the current resource.
    if kind == "same_resource":
        target_resource = current_resource
        target_index = resource_indexes[current_resource]

        matches = target_index.get(fragment, [])

        result["resolved_resource"] = target_resource
        result["target_count"] = len(matches)
        result["resolution_status"] = (
            "resolved_same_resource"
            if len(matches) == 1
            else "duplicate_target"
            if len(matches) > 1
            else "unresolved_same_resource"
        )

        return result

    # Project resource references.
    if kind == "project_resource":
        target_resource, lookup_status = resolve_resource_name(
            resource_name,
            exact_lookup,
            basename_lookup,
        )

        result["resolved_resource"] = target_resource
        result["resource_lookup_status"] = lookup_status

        if target_resource:
            matches = resource_indexes[target_resource].get(
                fragment, []
            )

            result["target_count"] = len(matches)
            result["resolution_status"] = (
                "resolved_project_resource"
                if len(matches) == 1
                else "duplicate_target"
                if len(matches) > 1
                else "unresolved_project_resource"
            )
        else:
            result["resolution_status"] = (
                "project_resource_not_in_archive"
            )

        return result

    # External URL/metamodel references.
    if kind == "external_url":
        result["resolution_status"] = "external_url_not_resolved"
        return result

    # Relative resource references, including other MDZIP files.
    if kind == "external_or_relative_resource":
        target_resource, lookup_status = resolve_resource_name(
            resource_name,
            exact_lookup,
            basename_lookup,
        )

        result["resolved_resource"] = target_resource
        result["resource_lookup_status"] = lookup_status

        if target_resource:
            matches = resource_indexes[target_resource].get(
                fragment, []
            )

            result["target_count"] = len(matches)
            result["resolution_status"] = (
                "resolved_relative_resource"
                if len(matches) == 1
                else "duplicate_target"
                if len(matches) > 1
                else "unresolved_relative_resource"
            )
        else:
            result["resolution_status"] = (
                "relative_resource_not_in_archive"
            )

        return result

    result["resolution_status"] = "unclassified_href"
    return result


def analyze_references(
    resources,
    resource_indexes,
    global_index,
    exact_lookup,
    basename_lookup,
    parent_maps,
):
    """
    Extract and resolve:
      - xmi:idref
      - href
      - candidate attributes matching known IDs
    """
    references = []

    for member, resource in resources.items():
        root = resource["root"]
        parent_map = parent_maps[member]
        local_index = resource_indexes[member]

        for element in root.iter():
            tag = local_name(element.tag)
            source_id, source_tag = nearest_identified_ancestor(
                element,
                parent_map,
            )

            # ------------------------------------------------
            # IDREF
            # ------------------------------------------------
            idref = get_attribute(element, "idref")

            if idref:
                matches = local_index.get(idref, [])

                references.append({
                    "source_resource": member,
                    "source_id": source_id,
                    "source_tag": source_tag,
                    "property_tag": tag,
                    "reference_kind": "idref",
                    "raw_value": idref,
                    "target_resource": (
                        member if len(matches) == 1 else None
                    ),
                    "target_id": idref,
                    "resolution_status": (
                        "resolved_same_resource"
                        if len(matches) == 1
                        else "duplicate_target"
                        if len(matches) > 1
                        else "unresolved_same_resource"
                    ),
                    "target_count": len(matches),
                })

            # ------------------------------------------------
            # HREF
            # ------------------------------------------------
            href = get_attribute(element, "href")

            if href:
                resolved = resolve_href(
                    href,
                    member,
                    resource_indexes,
                    exact_lookup,
                    basename_lookup,
                    global_index,
                )

                references.append({
                    "source_resource": member,
                    "source_id": source_id,
                    "source_tag": source_tag,
                    "property_tag": tag,
                    "reference_kind": "href",
                    "raw_value": href,
                    "target_resource": resolved.get(
                        "resolved_resource"
                    ),
                    "target_id": resolved.get("fragment"),
                    "resolution_status": resolved.get(
                        "resolution_status"
                    ),
                    "target_count": resolved.get("target_count", 0),
                    "resource_lookup_status": resolved.get(
                        "resource_lookup_status"
                    ),
                    "href_kind": resolved.get("href_kind"),
                })

            # ------------------------------------------------
            # ID-VALUED ATTRIBUTE CANDIDATES
            # ------------------------------------------------
            for attribute, value in element.attrib.items():
                attr_name = local_name(attribute)

                if attr_name in {"id", "ID", "idref", "href"}:
                    continue

                if attr_name not in REFERENCE_CANDIDATE_ATTRIBUTES:
                    continue

                # A QName such as uml:Property is not an ID.
                if ":" in value:
                    continue

                local_matches = local_index.get(value, [])
                global_matches = global_index.get(value, [])

                if not global_matches:
                    continue

                if len(local_matches) == 1:
                    status = "candidate_local_reference"
                    target_resource = member
                elif len(local_matches) > 1:
                    status = "candidate_duplicate_local_reference"
                    target_resource = None
                elif len(global_matches) == 1:
                    status = "candidate_cross_resource_reference"
                    target_resource = global_matches[0]["resource"]
                else:
                    status = "candidate_ambiguous_global_reference"
                    target_resource = None

                references.append({
                    "source_resource": member,
                    "source_id": source_id,
                    "source_tag": source_tag,
                    "property_tag": tag,
                    "reference_kind": f"attribute:{attr_name}",
                    "raw_value": value,
                    "target_resource": target_resource,
                    "target_id": value,
                    "resolution_status": status,
                    "target_count": len(global_matches),
                })

    return references


# ============================================================
# REPORTING
# ============================================================

def summarize(
    inventory,
    resources,
    resource_indexes,
    global_index,
    id_rows,
    references,
):
    print("\n" + "=" * 72)
    print("MDZIP ARCHIVE INVENTORY")
    print("=" * 72)

    print(f"Archive members:       {len(inventory)}")
    print(f"Parsed XML resources:  {len(resources)}")

    total_nodes = sum(
        resource["node_count"]
        for resource in resources.values()
    )

    print(f"Total XML nodes:       {total_nodes:,}")

    xml_success = sum(
        row["xml_parse_success"]
        for row in inventory
    )

    xml_candidates = sum(
        row["is_xml_candidate"]
        for row in inventory
    )

    print(f"XML candidates:        {xml_candidates}")
    print(f"XML parse successes:   {xml_success}")

    print("\n" + "=" * 72)
    print("IDENTIFIER INVENTORY")
    print("=" * 72)

    id_counts = Counter(
        row["source_id"] for row in id_rows
        if row["identifier_kind"] == "id"
    )

    duplicates_within_resource = []

    for member, index in resource_indexes.items():
        for source_id, elements in index.items():
            if len(elements) > 1:
                duplicates_within_resource.append(
                    (member, source_id, len(elements))
                )

    global_duplicates = {
        source_id: entries
        for source_id, entries in global_index.items()
        if len(entries) > 1
    }

    print(f"Elements with id:      {sum(id_counts.values())}")
    print(f"Unique IDs globally:   {len(id_counts)}")
    print(f"Duplicate IDs across resources: {len(global_duplicates)}")
    print(f"Duplicate IDs within resources: {len(duplicates_within_resource)}")

    print("\n" + "=" * 72)
    print("REFERENCE INVENTORY")
    print("=" * 72)

    reference_kinds = Counter(
        row["reference_kind"] for row in references
    )

    resolution_statuses = Counter(
        row["resolution_status"] for row in references
    )

    print("\nReference kinds:")
    for kind, count in reference_kinds.most_common():
        print(f"  {kind:38s} {count:>8,}")

    print("\nResolution statuses:")
    for status, count in resolution_statuses.most_common():
        print(f"  {status:38s} {count:>8,}")

    print("\n" + "=" * 72)
    print("RESOURCE SUMMARY")
    print("=" * 72)

    resource_summary = []

    for member, resource in resources.items():
        id_count = len(resource_indexes[member])

        ref_count = sum(
            1 for row in references
            if row["source_resource"] == member
        )

        resource_summary.append({
            "resource": member,
            "size_bytes": len(resource["data"]),
            "node_count": resource["node_count"],
            "unique_ids": id_count,
            "reference_occurrences": ref_count,
        })

    for row in sorted(
        resource_summary,
        key=lambda x: x["node_count"],
        reverse=True,
    )[:30]:
        print(
            f"{row['node_count']:>8,} nodes | "
            f"{row['unique_ids']:>6,} IDs | "
            f"{row['reference_occurrences']:>6,} refs | "
            f"{row['resource']}"
        )

    print("\n" + "=" * 72)
    print("UNRESOLVED / AMBIGUOUS REFERENCES")
    print("=" * 72)

    unresolved_statuses = {
        "unresolved_same_resource",
        "unresolved_project_resource",
        "project_resource_not_in_archive",
        "unresolved_relative_resource",
        "relative_resource_not_in_archive",
        "duplicate_target",
        "ambiguous_basename",
    }

    unresolved = [
        row for row in references
        if row["resolution_status"] in unresolved_statuses
    ]

    print(f"Potentially unresolved references: {len(unresolved)}")

    for row in unresolved[:MAX_EXAMPLES]:
        print(
            f"\n  {row['reference_kind']} | "
            f"{row['resolution_status']}"
        )
        print(f"    Source: {row['source_resource']}")
        print(f"    Element: {row['source_tag']} / {row['source_id']}")
        print(f"    Property: {row['property_tag']}")
        print(f"    Value: {row['raw_value']}")


# ============================================================
# MAIN
# ============================================================

def main():
    if not ARCHIVE_PATH.exists():
        raise FileNotFoundError(
            f"MDZIP archive not found: {ARCHIVE_PATH}\n"
            "Update ARCHIVE_PATH at the top of the script."
        )

    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    print(f"Reading archive: {ARCHIVE_PATH}")

    with zipfile.ZipFile(ARCHIVE_PATH, "r") as zip_file:
        inventory, resources = inventory_archive(zip_file)

    exact_lookup, basename_lookup = build_resource_lookup(
        resources
    )

    resource_indexes, global_index, id_rows = build_id_indexes(
        resources
    )

    parent_maps = build_parent_maps(resources)

    references = analyze_references(
        resources,
        resource_indexes,
        global_index,
        exact_lookup,
        basename_lookup,
        parent_maps,
    )

    summarize(
        inventory,
        resources,
        resource_indexes,
        global_index,
        id_rows,
        references,
    )

    # --------------------------------------------------------
    # CSV OUTPUTS
    # --------------------------------------------------------

    write_csv(
        OUTPUT_DIR / "archive_inventory.csv",
        inventory,
    )

    write_csv(
        OUTPUT_DIR / "resource_summary.csv",
        [
            {
                "resource": member,
                "size_bytes": len(resource["data"]),
                "node_count": resource["node_count"],
                "unique_ids": len(resource_indexes[member]),
            }
            for member, resource in resources.items()
        ],
    )

    write_csv(
        OUTPUT_DIR / "element_ids.csv",
        id_rows,
    )

    write_csv(
        OUTPUT_DIR / "references.csv",
        references,
    )

    duplicate_rows = []

    for source_id, entries in global_index.items():
        if len(entries) > 1:
            for entry in entries:
                duplicate_rows.append({
                    "source_id": source_id,
                    "resource": entry["resource"],
                    "tag": local_name(entry["element"].tag),
                    "name": get_attribute(
                        entry["element"], "name"
                    ),
                    "duplicate_count": len(entries),
                })

    write_csv(
        OUTPUT_DIR / "duplicate_ids.csv",
        duplicate_rows,
        fieldnames=[
            "source_id",
            "resource",
            "tag",
            "name",
            "duplicate_count",
        ],
    )

    unresolved = [
        row for row in references
        if row["resolution_status"] not in {
            "resolved_same_resource",
            "resolved_project_resource",
            "resolved_relative_resource",
            "external_url_not_resolved",
            "candidate_local_reference",
            "candidate_cross_resource_reference",
        }
    ]

    write_csv(
        OUTPUT_DIR / "unresolved_references.csv",
        unresolved,
    )

    # Save a compact machine-readable summary.
    summary = {
        "archive": str(ARCHIVE_PATH),
        "archive_members": len(inventory),
        "parsed_xml_resources": len(resources),
        "total_xml_nodes": sum(
            resource["node_count"]
            for resource in resources.values()
        ),
        "unique_global_ids": len(global_index),
        "duplicate_global_ids": sum(
            1 for entries in global_index.values()
            if len(entries) > 1
        ),
        "reference_count": len(references),
        "reference_kinds": dict(
            Counter(row["reference_kind"] for row in references)
        ),
        "resolution_statuses": dict(
            Counter(row["resolution_status"] for row in references)
        ),
    }

    with (OUTPUT_DIR / "summary.json").open(
        "w", encoding="utf-8"
    ) as f:
        json.dump(summary, f, indent=2)

    print("\n" + "=" * 72)
    print("REPORTS WRITTEN")
    print("=" * 72)

    for path in sorted(OUTPUT_DIR.iterdir()):
        print(f"  {path}")


if __name__ == "__main__":
    main()
````
