# Graph Connectivity:

A graph $G = (N,E)$ is connected if there exists a path for all $n_i \rightarrow n_j \in N$.

## Seperation:

Node sets $A, B \subseteq N: A\cup B=N$ are seperable if no edge connects $A\backslash B$ to $B\backslash A \Rightarrow S = A\cap B$ is a seperating set. 

## Connected:

$G$ is $k$ connected if there is not proper seperation of order less than $k$
